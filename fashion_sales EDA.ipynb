```python
df.loc[df['category'] == 'Accessories', 'size'] = df['size'].fillna('Not Applicable') #it means that if the category is Accessories then the size will be filled with Not Applicable
```

```python
df.loc[df['category'] != 'Accessories', 'size'] = df['size'].fillna('Unknown') #it means that if the category is not Accessories then the size will be filled with Unknown
```
